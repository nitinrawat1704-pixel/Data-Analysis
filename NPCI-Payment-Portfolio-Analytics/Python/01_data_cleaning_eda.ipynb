```python
import pandas as pd
import numpy as np

customers=pd.read_csv("/content/customers.csv")
merchants=pd.read_csv("/content/merchants.csv")
offer_usage=pd.read_csv("/content/offer_usage.csv")
offers=pd.read_csv("/content/offers.csv")
transactions=pd.read_csv("/content/transactions.csv")



```

```python
#cleaning customers dataset

cdf=pd.DataFrame(customers)

# cdf.isnull().sum()                                 #finding column with null
# print(cdf["income"].isnull().sum())
cdf["income"]=cdf["income"].fillna(cdf["income"].mean().round())  #replacing null with that columns avg
# print(cdf["income"].isnull().sum())

# print(cdf.duplicated().sum())                                #checking for duplicate values

cdf["join_date"]=pd.to_datetime(cdf["join_date"],errors="coerce")  #changed it data type to date
```

```python
#cleaning merchants dataset

mdf=pd.DataFrame(merchants)
mdf["onboarding_date"]=pd.to_datetime(mdf["onboarding_date"],errors="coerce")
mdf.info()

print(mdf.isnull().sum())
print(mdf.duplicated().sum())
print(mdf["merchant_id"].duplicated().sum())
print(mdf["category"].value_counts())
print(mdf["merchant_size"].value_counts())
print(mdf["city"].value_counts())
```

```python
#cleaning transactions dataset


tdf=pd.DataFrame(transactions)
# print(tdf)

tdf.isnull().sum()
tdf["amount"]=tdf["amount"].fillna(tdf["amount"].mean().round())  #replaced null value with avg

tdf.duplicated().sum()


print(tdf[tdf["amount"]<0])         #there are 50 rows with negative transaction and i'll delete them
tdf=tdf[tdf["amount"]>=0]
print(tdf[tdf["amount"]<0])         #its gone

tdf["transaction_date"]=pd.to_datetime(tdf["transaction_date"],errors="coerce")  #changed the data type to date


tdf["transaction_status"].value_counts()
```

```python
# cleaning offer_usage dataset

oudf=pd.DataFrame(offer_usage)
# oudf.info()

print(oudf["transaction_id"].duplicated().sum())

print(oudf.isnull().sum())
oudf["reward_amount"]=oudf["reward_amount"].fillna(oudf["reward_amount"].mean().round())   #replaced null val with avg
print(oudf.isnull().sum())

```

```python
# cleaning offers dataset

odf=pd.DataFrame(offers)

odf.info()
odf["start_date"]=pd.to_datetime(odf["start_date"])    #change the data type
odf["end_date"]=pd.to_datetime(odf["end_date"])        #change the data type

odf.isnull().sum()

print(odf[odf["offer_cost"]<0])                       #checking for negative
print(odf[odf["reward_percentage"]<0])



odf.duplicated().sum()                    #checking for duplicates

print(odf[odf["reward_percentage"]>100])    # checking for buisness logic
print(odf[odf["start_date"]>odf["end_date"]]) # checking for buisness logic
```

```python


# Which payment product generates the highest successful transaction value?

product_analysis=tdf[tdf["transaction_status"]=="Success"].groupby("payment_product").agg(
                                                                        transaction_count=("transaction_id","count"),
                                                                        transaction_value=("amount","sum"),
                                                                        avg_transaction=("amount","mean")
).sort_values("transaction_value",ascending=False)

print(product_analysis)

# Which cities generate the highest transaction value?

tdf.merge(mdf[["merchant_id", "city", "category", "merchant_size"]],on="merchant_id",how="left").groupby("city")["amount"].sum()
```
